```python
import tensorflow as tf
import numpy as np
from PIL import Image

model = tf.keras.applications.MobileNetV3Small(
    input_shape=(224, 224, 3),
    weights="imagenet"
)

print("Model ready")
```

```python
def predict_image(path):
    image = Image.open(path).convert("RGB")
    image = image.resize((224, 224))

    x = np.array(image, dtype=np.float32)
    x = tf.keras.applications.mobilenet_v3.preprocess_input(x)
    x = np.expand_dims(x, axis=0)

    prediction = model.predict(x, verbose=0)

    results = tf.keras.applications.mobilenet_v3.decode_predictions(
        prediction,
        top=6
    )[0]

    return [
        {
            "class": name,
            "confidence": float(score)
        }
        for _, name, score in results
    ]
```

```python
results = predict_image("cat.jpg")

for r in results:
    print(r["class"], "=", r["confidence"])
```

```python
leaky_response = predict_image("cat.jpg")

print(leaky_response)
```

```python
protected_response = {
    "class": leaky_response[0]["class"]
}

print(protected_response)
```
